# **Question 3**

**Download required packages and import libraries**

```python
import nltk
nltk.download('movie_reviews')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')

from nltk.corpus import movie_reviews, stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer
from nltk.probability import FreqDist
import matplotlib.pyplot as plt
```

**View the list of negative review files and select one text**

```python
neg_files = movie_reviews.fileids('neg')
print(neg_files[:5])

text = movie_reviews.raw(neg_files[0])
print(text[:1000])
```

**Tokenize the text into words**

```python
tokens = word_tokenize(text.lower())
print(tokens[:50])
```

**Remove stopwords and punctuation**

```python
stop_words = stopwords.words('english')

filtered_words = []
for w in tokens:
  if w not in stop_words and w.isalpha():
    filtered_words.append(w)

print(filtered_words[:50])
```

**Apply lemmatization to the filtered words**

```python
lemmatizer = WordNetLemmatizer()

lemmatized_words = []
for w in filtered_words:
  lemmatized_words.append(lemmatizer.lemmatize(w))

print(lemmatized_words[:50])
```

**Identify the 20 most frequent words**

```python
word_count = FreqDist(lemmatized_words)
top_20 = word_count.most_common(20)
print(top_20)
```

**Visualize the 20 most frequent words using a bar plot**

```python
word = []
count = []
for tup in word_count.most_common(20):
  word.append(tup[0])
  count.append(tup[1])

plt.figure(figsize = (12, 6))
plt.bar(word, count)
plt.title("Top 20 Most Frequent Words in Negative Movie Review")
plt.xlabel("Words")
plt.ylabel("Frequency")
plt.xticks(rotation = 45)
plt.show()
```

# **Question 4**

```python
from nltk.util import ngrams
import numpy as np
from tabulate import tabulate
```

```python
file_list = reuters.fileids()
print(file_list[:5])

text = reuters.raw(file_list[0])
print(text[:1000])
```

```python
stop_words = stopwords.words('english')
tokens = word_tokenize(text.lower())

cleaned_words = []
for w in tokens:
  if w.isalpha() and w not in stop_words:
    cleaned_words.append(w)

print(cleaned_words[:30])
```

```python
word_count = FreqDist(cleaned_words)
top_12_tuples = word_count.most_common(12)

top_12_words = []
for tup in top_12_tuples:
  top_12_words.append(tup[0])

print(top_12_words)
```

```python
bigrams_list = list(ngrams(cleaned_words, 2))
print(bigrams_list[:15])
```

```python
matrix = np.zeros((len(top_12_words), len(top_12_words)), dtype=int)

word_to_index = {}
for i, word in enumerate(top_12_words):
  word_to_index[word] = i

for pair in bigrams_list:
  w1 = pair[0]
  w2 = pair[1]
  if w1 in word_to_index and w2 in word_to_index:
    idx1 = word_to_index[w1]
    idx2 = word_to_index[w2]
    matrix[idx1, idx2] += 1

print(matrix)
```

```python
table_data = []
for i, word in enumerate(top_12_words):
  row = [word] + list(matrix[i])
  table_data.append(row)

headers = ["Word"] + top_12_words
print(tabulate(table_data, headers=headers, tablefmt="grid"))
```

# **Question 5**

**Necessary imports**

```python
import pandas as pd
import re
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, accuracy_score
```

**Mount Drive**

**Load the IMDB dataset**

```python
df = pd.read_csv('/content/drive/MyDrive/IMDB Dataset/IMDB Dataset.csv')

print(df.head())
print(df.shape)
```

**Select the first 40,000 rows**

```python
df = df.head(40000).copy()

print(df.shape)
```

**Preprocess the review texts**

```python
import re

def clean_text(text):
    text = text.lower()
    text = re.sub(r'<br\s*/?>', ' ', text)
    text = re.sub(r'[^\w\s]', '', text)
    return text
```

**Apply text preprocessing**

```python
df['review'] = df['review'].apply(clean_text)

print(df['review'].iloc[0][:300])
```

**Encode the sentiment column**

```python
df['sentiment'] = df['sentiment'].map({'negative': 0, 'positive': 1})

print(df['sentiment'].value_counts())
```

**Split the dataset into training and testing sets**

```python
from sklearn.model_selection import train_test_split

X = df['review']
y = df['sentiment']

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.25,
    random_state=42,
    stratify=y
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

**Convert the text into Bag-of-Words features**

```python
from sklearn.feature_extraction.text import CountVectorizer

vectorizer = CountVectorizer(max_features=10000)

X_train_bow = vectorizer.fit_transform(X_train)
X_test_bow = vectorizer.transform(X_test)
```

**Display the dimensions of the feature matrices**

```python
print("Training feature matrix:", X_train_bow.shape)
print("Testing feature matrix:", X_test_bow.shape)
```

# **Train a logistic regression and random forest classifier using the feature vectors**

**Train the Logistic Regression classifier**

```python
from sklearn.linear_model import LogisticRegression

logistic_model = LogisticRegression(max_iter=1000)

logistic_model.fit(X_train_bow, y_train)
```

**Evaluate Logistic Regression**

```python
from sklearn.metrics import classification_report, accuracy_score

logistic_predictions = logistic_model.predict(X_test_bow)

print(classification_report(
    y_test,
    logistic_predictions,
    target_names=['negative', 'positive']
))

print("Accuracy:", accuracy_score(y_test, logistic_predictions))
```

**Train the Random Forest classifier**

```python
from sklearn.ensemble import RandomForestClassifier

rf_model = RandomForestClassifier(
    n_estimators=100,
    max_depth=20,
    random_state=42,
    n_jobs=-1
)

rf_model.fit(X_train_bow, y_train)
```

**Evaluate Random Forest**

```python
rf_predictions = rf_model.predict(X_test_bow)

print(classification_report(
    y_test,
    rf_predictions,
    target_names=['negative', 'positive']
))

print("Accuracy:", accuracy_score(y_test, rf_predictions))
```

**Compare both classifiers**

```python
logistic_accuracy = accuracy_score(y_test, logistic_predictions)
rf_accuracy = accuracy_score(y_test, rf_predictions)

print("Logistic Regression Accuracy:", logistic_accuracy)
print("Random Forest Accuracy:", rf_accuracy)
```

**Compare model accuracy using a bar chart**

```python
import matplotlib.pyplot as plt

models = ['Logistic Regression', 'Random Forest']
accuracies = [logistic_accuracy * 100, rf_accuracy * 100]

plt.bar(models, accuracies)
plt.ylabel('Accuracy (%)')
plt.title('Model Accuracy Comparison')
plt.ylim(0, 100)
plt.show()
```

**Visualize the Logistic Regression confusion matrix**

```python
from sklearn.metrics import ConfusionMatrixDisplay

ConfusionMatrixDisplay.from_predictions(
    y_test,
    logistic_predictions,
    display_labels=['Negative', 'Positive']
)

plt.title('Logistic Regression Confusion Matrix')
plt.show()
```

**Random Forest Confusion Matrix**

```python
ConfusionMatrixDisplay.from_predictions(
    y_test,
    rf_predictions,
    display_labels=['Negative', 'Positive']
)

plt.title('Random Forest Confusion Matrix')
plt.show()
```
